## 1. Import Libraries

```python
import pandas as pd
import numpy as np
```

## 2. Load Raw Data

```python
df = pd.read_csv('Pay-sim.csv')
print('Shape:',df.shape)
df.head()
```

## 3. Remove Zero-Amount Rows

```python
df = df[df['amount'] > 0].reset_index(drop=True)
print('After removing zeros:',df.shape)
```

## 4. Remove Duplicates

```python
df = df.drop_duplicates().reset_index(drop=True)
print('After removing duplicates:',df.shape)
```

## 5. Extract isDestMerchant

```python
df['isDestMerchant'] = (df['nameDest'].str[0] == 'M').astype(int)
print(df['isDestMerchant'].value_counts())
```

## 6. Drop ID & Unreliable Columns

```python
df.drop(columns=['nameOrig','nameDest','isFlaggedFraud'], inplace=True)
print(df.columns.tolist())
```

## 7. Encode Transaction Type

```python
type_map = {'PAYMENT':0,'TRANSFER':1,'CASH_OUT':2,'CASH_IN':3,'DEBIT':4}
df['type'] = df['type'].map(type_map)
print(df['type'].value_counts())
```

## 8. Verify — No Nulls, No Duplicates

```python
print('Nulls:',df.isnull().sum().sum())
print('Duplicates:',df.duplicated().sum())
df.describe()
```

## 9. Save Cleaned Data

```python
df.to_csv('Pay-sim_cleaned.csv', index=False)
print('Saved -> Pay-sim_cleaned.csv')
print('Final shape:',df.shape)
```
