## 1. Import Libraries

```python
import pandas as pd
import numpy as np
```

## 2. Load Cleaned Data

```python
df = pd.read_csv('Pay-sim_cleaned.csv')
print('Shape:',df.shape)
df.head()
```

## 3. Engineer Balance Error Features

```python
df['errorBalanceOrig'] = df['newbalanceOrig'] + df['amount'] - df['oldbalanceOrg']
df['errorBalanceDest'] = df['oldbalanceDest'] + df['amount'] - df['newbalanceDest']
print(df[['errorBalanceOrig','errorBalanceDest']].describe())
```

## 4. Winsorize Outliers (cap at 1st–99th percentile)

```python
monetary = ['amount','oldbalanceOrg','newbalanceOrig','oldbalanceDest','newbalanceDest','errorBalanceOrig','errorBalanceDest']
for col in monetary:
    p01 = df[col].quantile(0.01)
    p99 = df[col].quantile(0.99)
    df[col] = df[col].clip(lower=p01, upper=p99)
print('Winsorization done')
```

## 5. Log1p Transform (reduce skewness)

```python
for col in monetary:
    df[f'log_{col}'] = np.log1p(df[col].clip(lower=0))
print('Log features created:', [f'log_{c}' for c in monetary])
```

## 6. Final Feature Check

```python
print('Shape:',df.shape)
print('Columns:',df.columns.tolist())
print('Nulls:',df.isnull().sum().sum())
```

## 7. Save Feature-Engineered Data

```python
df.to_csv('Pay-sim_features.csv', index=False)
print('Saved -> Pay-sim_features.csv')
```
